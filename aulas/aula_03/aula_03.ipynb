<h2 style="color:#8B0000;">📖 Como Funciona a Aplicação Web?</h2>

Neste projeto da <b>Aula 03</b>, avançamos da execução de scripts locais (em janelas do Sistema Operacional) para a <b>Web</b>. Construímos uma aplicação onde a captura de vídeo acontece através do navegador do usuário (Frontend), enviando os frames de forma extremamente rápida para o nosso Servidor Python (Backend) fazer a inferência.

<h3 style="color:#B22222;">🔧 As Ferramentas que Vamos Usar</h3>
<ul>
<li><b>FastHTML / Starlette:</b> Framework minimalista que nos permite criar telas e configurar rotas WebSocket na mesma estrutura de forma fácil.</li>
<li><b>WebSockets:</b> Protocolo fundamental desta aplicação. Oferece comunicação bidirecional contínua de baixíssima latência (útil para streaming de vídeo em tempo real).</li>
<li><b>MediaPipe (Google):</b> Utilizado no servidor para localizar e rastrear 21 pontos (<i>Landmarks</i>) fundamentais das mãos dentro da imagem recebida computacionalmente.</li>
<li><b>Scikit-Learn:</b> Aplicamos nosso modelo treinado (<i>SVM/RandomForest</i>) para classificar os landmarks e "adivinhar" o gesto atual que o usuário fez em frente a Webcam.</li>
</ul>

<h3 style="color:#B22222;">⚙️ O Pipeline e Fluxo de Dados</h3>
<ol>
<li>O <b>Navegador (Frontend)</b> inicializa a webcam e captura frames que são projetados em um <code>&lt;canvas&gt;</code> HTML.</li>
<li>Cada imagem projetada é imediatamente serializada e os binários/bytearrays viajam pelo <b>WebSocket</b> para o Servidor.</li>
<li>O <b>Servidor (Backend Python em FastHTML)</b> captura esses bytes na Rota <code>/ws</code>.</li>
<li>O OpenCV reconstrói os bytes transformando-os numa imagem RGB e joga para o <i>HandLandmarker</i>.</li>
<li>Com a posição das mãos identificadas, o <b>Modelo de ML</b> faz a previsão do gesto correspondente (se a mão estiver em cena).</li>
<li>O Back-End joga instantaneamente essa detecção de texto (como "Desconhecido", ou "Gesto_X") de volta pelo duto do WebSocket.</li>
<li>O Java Script do navegador recebe o texto, pinta essa resposta com cor sólida e repete o processo infinitamente!</li>
</ol>

<div style="padding:14px 18px; border-radius:8px; border-left:4px solid #B22222; margin:10px 0;">
💡 <b>Por que WebSockets e não requisições comuns?</b> Um vídeo é na verdade dúzias de fotos p/ segundo. Requisitar a cada décimo de segundo via HTTP REST o peso de re-inicializar novos "handshakes" geraria um buffer insano. WebSockets ignoram esse tempo por criarem um "cano virtual" fluído e constante!
</div>

<h2 style="color:#8B0000;">📁 Estrutura Refatorada do Projeto</h2>

Projetos Web demandam uma melhor organização de diretórios do que Jupyter Notebooks nativos. Ao invés da lógica num script monolítico, nossa árvore atual obedece separação de responsabilidades:

```python
import os

def print_tree(startpath, level=0):
    for entry in os.listdir(startpath):
        if entry in ['__pycache__', '.ipynb_checkpoints'] or entry.endswith('.pyc'):
            continue
        path = os.path.join(startpath, entry)
        print('    ' * level + '├── ' + entry)
        if os.path.isdir(path):
            print_tree(path, level + 1)

print('aulas/aula_03/')
print_tree('.')
```
